### What it means for Nevis

1. **Core market:** the 1,409 wealth-focused RIAs at <span>&#36;</span>1B–<span>&#36;</span>10B. They are scaled enough to buy software and independent enough to decide fast, and the band grew +27% in firm count since 2021.
2. **The pitch is the capacity ceiling:** wealth firms grow by hiring advisors, not by making each advisor more productive. Raising per-advisor capacity is the value Nevis sells.
3. **Prioritize** firms that hit two or more of: consistent top-quartile growth (Q1), AUM per advisor above the segment plateau (Q2), and a persistent staffing gap (Q3).
4. **Watch list:** 115 digital platforms serve 25.6M clients with ~3,056 advisors (`outputs/digital_platforms.csv`), the competitive benchmark for what technology does to advisor capacity. A further 52 investment managers used by RIAs (<span>&#36;</span>450.2B of AUM; `outputs/investment_managers.csv`) are potential partners or channels.

### Scope: why the rankings focus on wealth-focused RIAs (for now)

- **Nevis sells to wealth managers.** Its AI platform supports advisors who look after individual clients: meetings, planning, reporting and service. Wealth-focused RIAs are the firms doing that work, so they are the buyers.
- **The capacity questions only make sense for advice businesses.** AUM and clients per advisor measure how much each advisor carries when advisors serve clients personally. A fund manager with <span>&#36;</span>1B per advisor and two clients is not capacity-strained, so mixing firm types would make the Q2 and Q3 rankings meaningless.
- **Some firms look like wealth managers in the data but aren't.** Digital platforms serve clients through technology rather than advisors. Investment managers used by RIAs report other advisers' clients as their own. Family offices serve a handful of very wealthy families, so their AUM per advisor reflects a few large relationships rather than strain. None of these is a buyer of advisor-productivity tools in the usual sense, so all three are excluded by rule.
- **"For now" means nothing is thrown away.** Every ranking also has an all-RIA list, and the full firm table keeps every firm with its type flags. The scope can be widened later, for example to hybrid advisors or institutional consultants, by changing one filter.

How every 2026 firm is classified (rules applied top to bottom; a firm gets one type):

| Firm type | Rule | Firms | AUM |
|---|---|---:|---:|
| Digital platform (excluded) | More than 1,000 clients per advisor: robo-advisers, app-based investing, managed-account platforms (e.g. Betterment, Acorns, AssetMark) | 115 | <span>&#36;</span>7.3T |
| Investment manager used by RIAs (excluded) | ≥ 50% individual AUM, but offers no financial planning (Item 5G(1), 5H) and does not select other advisers (5G(7)), and either (a) ≥ 25% of AUM managed as a sub-adviser in other firms' wrap programs (5I(2)(b)), or (b) more than 200 clients per advisor with at least 10 advisors and at least <span>&#36;</span>250K AUM per client (e.g. Quantinno, Vise AI, Aperio) | 52 | <span>&#36;</span>450.2B |
| Family office (excluded) | ≥ 50% individual AUM, fewer than 20 clients and at least <span>&#36;</span>100M AUM: a handful of very wealthy families (median ~<span>&#36;</span>26M per client), e.g. BT Family Office, Fifth Avenue Family Office | 163 | <span>&#36;</span>93.3B |
| Wealth-focused RIA (the focus of the rankings) | ≥ 50% of AUM from individuals and high-net-worth individuals (Item 5D(a)+(b)), and not one of the types above | 8,899 | <span>&#36;</span>23.3T |
| <50% individual AUM (shown only in all-RIA lists) | Everything else: mostly fund, ETF, pension and other institutional managers | 7,920 | <span>&#36;</span>145.8T |

These thresholds are judgment calls, set where the data shows natural breaks (Section 2.1 explains each one). The tests:
- **Planning and adviser selection** separate managers from advisors: every platform-style firm checked answers 'no' to both, and traditional advisors answer 'yes'.
- **The 10-advisor and <span>&#36;</span>250K-per-client conditions** stop high client volume alone from excluding real advisory firms. Mass-affluent and bank-channel advisers (e.g. Nest Egg, about <span>&#36;</span>47K per client) and one-person investment-counsel boutiques stay wealth-focused.
- **The family-office rule needs at least <span>&#36;</span>100M of AUM** (the SEC registration threshold). Firms with under 20 clients and less AUM are small practices, so they stay wealth-focused; they are only kept out of the capacity ranking (Q2), which requires at least 20 clients.
- **Known edge cases:** Edelman Financial Engines (human planners plus a large 401(k) advice business) is excluded by the 1,000-client platform rule. Asset managers that also report planning (e.g. GW&K) stay wealth-focused.

### Key assumptions

| Topic | Assumption |
|---|---|
| Universe | SEC-registered investment advisers only. State-registered advisers (mostly < <span>&#36;</span>100M) and exempt reporting advisers are not in these files |
| Time | Six September snapshots. AUM is self-reported as of each firm's fiscal year-end (mostly the prior December), so growth includes market returns, not only net new money |
| AUM | Regulatory AUM, Form ADV Item 5F(2)(c) |
| Advisors | Item 5B(1): employees performing investment advisory functions, research included. Form ADV has no 'advisor' field. Re-running with 5B(3), state-registered representatives, keeps the results: rank correlation 0.92–0.96 for wealth-focused firms (0.79–0.95 across all RIAs), though some individual top-10 names change |
| Clients | Σ Item 5D client counts, with 'fewer than 5' checkboxes counted as 2. A second measure adds Item 5C(1) clients with no managed assets. Obvious entry errors are removed |
| Segments | Latest-year AUM, half-open bands: [0, <span>&#36;</span>1B), [<span>&#36;</span>1B, <span>&#36;</span>10B), [<span>&#36;</span>10B, <span>&#36;</span>50B), [<span>&#36;</span>50B, ∞). Growth tables group firms by their latest-year segment |
| Growth | CAGR between the start snapshot and the latest one. Only firms present in both years with ≥ <span>&#36;</span>100M AUM at the start and current (non-stale) filings are eligible, so tiny bases do not produce meaningless rates |
| Firm types | Rankings focus on wealth-focused RIAs. Digital platforms, investment managers used by RIAs and family offices are excluded by rule; see the Scope section above for the exact rules, the counts and the rationale |
| Ratios | AUM per advisor needs ≥ 2 advisors; the wealth-focused capacity ranking also needs ≥ 20 clients. Under-staffing needs ≥ 2 advisors at the start. A 'material' gap is ≥ 10 pts/yr of AUM growth over advisor growth |
| Stale filings | A firm whose latest ADV is more than 15 months older than the snapshot missed its annual update, so its AUM is a repeat; it is excluded from growth and ratio rankings (~0.4% of firms a year) |
| Caveats | M&A inflates growth and is flagged only partly (`consolidation_flag`). Firms that exit between snapshots are not measured. A drop in 5B(1) can be a re-classification rather than attrition. AUM totals double-count assets that one adviser sub-advises for another, so market totals and segment AUM shares are somewhat overstated |

These are the key assumptions only. The **complete list (16 items), the full methodology, every threshold and the output data dictionary are in `README.md`** (§4–§7). The raw-data mapping is in `docs/field_mapping.md`.

```python
# ---- display helpers ---------------------------------------------------------
# Jupyter renders text between two "$" signs as LaTeX, which garbles tables such as "$1B-$10B".
pd.set_option("display.html.use_mathjax", False)

# Readable column / index names for every displayed table (the CSVs keep the raw names).
LABELS = {
    "sec_number": "SEC #", "firm_name": "Firm", "legal_name": "Legal name", "website": "Website", "city": "City",
    "state": "State", "segment": "Segment", "rank": "Rank", "year": "Year", "filing_date": "Filed",
    "aum": "AUM", "aum_start": "AUM (start)", "aum_2021": "AUM 2021", "aum_cagr": "AUM CAGR",
    "aum_growth_total": "AUM growth (total)", "advisors": "Advisors", "advisors_start": "Advisors (start)",
    "advisors_2021": "Advisors 2021", "advisor_cagr": "Advisor CAGR", "growth_gap": "AUM CAGR − advisor CAGR",
    "aum_per_advisor": "AUM / advisor", "aum_per_advisor_start": "AUM / advisor (start)",
    "clients_raum": "Clients", "clients_total": "Clients incl. non-AUM", "aum_per_client": "AUM / client",
    "clients_per_advisor": "Clients / advisor", "employees": "Employees", "pct_aum_individuals": "Individual AUM",
    "consolidation_flag": "M&A flag", "firm_type": "Firm type",
    "firms": "Firms", "total_aum": "Total AUM", "median_aum": "Median AUM", "median_advisors": "Median advisors",
    "median_aum_per_advisor": "Median AUM / advisor", "median_clients": "Median clients",
    "wealth_focused_firms": "Wealth-focused firms", "digital_platforms": "Digital platforms",
    "investment_managers": "Investment managers", "family_offices": "Family offices",
    "share_of_firms": "Share of firms", "share_of_aum": "Share of AUM", "wealth_focused_share": "Wealth-focused share",
    "aum_zero": "AUM = 0", "advisors_zero": "Advisors = 0", "fewer_than_5_imputed": "'<5 clients' imputed",
    "c5_invalid": "5C(1) errors", "stale_filings": "Stale filings", "location_filled": "Location filled",
    "location_missing": "Location missing", "website_borrowed": "Website from other year",
    "website_social_only": "Social-only website", "website_missing": "Website missing",
    "continuing": "Continuing", "entered": "Entered", "exited": "Exited",
}

def money(x):
    if pd.isna(x):
        return "–"
    for div, suffix in [(1e12, "T"), (1e9, "B"), (1e6, "M"), (1e3, "K")]:
        if abs(x) >= div:
            return f"${x / div:,.1f}{suffix}"
    return f"${x:,.0f}"

def pct(x, signed=True):
    return "–" if pd.isna(x) else (f"{x:+.1%}" if signed else f"{x:.0%}")

def fmt(df, money_cols=(), pct_cols=(), share_cols=(), int_cols=(), flag_cols=()):
    """Format numbers for reading and apply readable labels; the unformatted data goes to outputs/."""
    d = df.copy()
    for c in money_cols:
        d[c] = d[c].map(money)
    for c in pct_cols:
        d[c] = d[c].map(pct)
    for c in share_cols:
        d[c] = d[c].map(lambda x: pct(x, signed=False))
    for c in int_cols:
        d[c] = d[c].map(lambda x: "–" if pd.isna(x) else f"{x:,.0f}")
    for c in flag_cols:
        d[c] = d[c].map(lambda x: "Yes" if x is True or x == 1 else "")
    if "firm_name" in d.columns:
        d["firm_name"] = d["firm_name"].map(ap.display_name)
    for c in d.columns:  # missing text (e.g. no state for non-US firms) shows as "–", not "NaN"
        if d[c].isna().any():
            d[c] = d[c].astype(object).where(d[c].notna(), "–")
    d = d.rename(columns=LABELS)
    d.index = d.index.set_names([LABELS.get(n, n) for n in d.index.names])
    d.columns = d.columns.set_names([LABELS.get(n, n) for n in d.columns.names])
    return d

def show(df):
    """Display a formatted table as a plain DataFrame. Jupyter: use_mathjax=False (above) keeps "$1B-$10B"
    from being read as LaTeX. PyCharm/VS Code show DataFrames in their own table viewer, which displays the
    cell text as-is, so no HTML tricks are used here (they would show up as literal tags)."""
    display(df)

def per_segment(counts):
    """'<$1B: 477 · $1B-$10B: 340 · ...' instead of a raw dict."""
    return " · ".join(f"{k}: {v:,}" for k, v in counts.items())

def heading(text, level=4):
    """Markdown heading inside a code cell. Each "$" gets its own <span> so no renderer pairs them as LaTeX."""
    text = text.replace("<$", "&lt;$").replace(">$", "&gt;$")
    display(Markdown("#" * level + " " + text.replace("$", "<span>&#36;</span>")))

def ranked_lists(df, sort_col, cols, title, wealth_min_clients=0, csv_note=None, **fmt_kw):
    """Top-N per segment for wealth-focused RIAs (Nevis's core market), then for all RIAs.
    Displays TOP_N wealth-focused and 5 overall; both lists are returned at TOP_N for the CSV.
    wealth_min_clients applies a client floor to the wealth-focused list only."""
    out = []
    wealth = df[df["wealth_focused"] & (df["clients_raum"] >= wealth_min_clients)]
    for label, d, n_show in [("Wealth-focused RIAs", wealth, TOP_N), ("All RIAs", df, 5)]:
        top = ap.top_n_by_segment(d, sort_col, TOP_N, cols)
        out.append(top.assign(universe=label))
        note = "" if n_show == TOP_N else f" (top {n_show} shown, {csv_note or f'top {TOP_N} in CSV'})"
        heading(f"{title}: {label}{note}")
        show(fmt(top[top["rank"] <= n_show].set_index(["segment", "rank"]), **fmt_kw))
    return pd.concat(out)

# ---- chart style (one blue for single series; 3 validated hues for horizons) --
BLUE, ORANGE, AQUA = "#2a78d6", "#eb6834", "#1baf7a"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#e4e3df"
plt.rcParams.update({
    "figure.dpi": 110, "font.size": 10, "axes.edgecolor": GRID, "axes.labelcolor": MUTED,
    "axes.titlesize": 11, "axes.titleweight": "bold", "axes.titlecolor": INK,
    "axes.spines.top": False, "axes.spines.right": False, "axes.grid": True,
    "grid.color": GRID, "grid.linewidth": 0.8, "xtick.color": MUTED, "ytick.color": MUTED,
    "legend.frameon": False, "axes.axisbelow": True,
    "text.parse_math": False,   # "$1B-$10B" must not be read as LaTeX
})
```

## 2. Build the firm-year table

**Sources:** six SEC "Information about Registered Investment Advisers" snapshots, one per September 2021–2026.
They come in two layouts: 262 columns (2021–22 CSV) and 448 columns (2023–25 XLSX and the 2026 roster CSV).
Column *names* are identical across layouts, so `adv_pipeline` maps by name and never by position.
It also handles three date encodings, comma-formatted numbers, and cp1252 text in the 2026 file.

**Key:** `Organization CRD#` is the firm's permanent identifier and is unique within each snapshot. It links firms across years.

**Field choices** (full definitions: README §4.2; raw-column mapping: `docs/field_mapping.md`):

| Field | Form ADV item | Note |
|---|---|---|
| AUM | 5F(2)(c) | Total regulatory AUM |
| Advisors | 5B(1) | Employees performing investment advisory functions (incl. research). 5B(3), state-registered IARs, kept for a sensitivity check |
| Clients (`clients_raum`) | Σ 5D(a..n)(1) | Client counts by type. Each "fewer than 5 clients" checkbox imputed as 2 |
| Clients (`clients_total`) | `clients_raum` + 5C(1) | Adds clients the firm advises without managing assets (e.g. planning-only) |
| Employees | 5A | |

### 2.1 Firm types: who counts as a wealth-focused RIA

**Why the rankings focus on wealth-focused RIAs (for now).**
- **Buyers:** Nevis sells an AI platform to wealth managers, firms whose advisors look after individual clients.
- **Meaningful metrics:** capacity measures such as AUM and clients per advisor only mean "strain" in an advice business.
- **Look-alikes:** some firms look like wealth managers in the data but aren't advisor practices:
  - digital platforms serve clients through technology;
  - investment managers serve them through other advisers;
  - family offices serve a handful of very wealthy families.

  These are partners, channels, competitors or a different sale rather than buyers of advisor-productivity tools.

Every ranking therefore leads with wealth-focused firms and keeps an all-RIA list alongside. The full firm table carries all
the type flags, so the scope can be widened later by changing one filter.

The rules are applied in this order, and each firm gets exactly one type:

| Type | Rule | Why |
|---|---|---|
| `digital_platform` (excluded) | > 1,000 clients per advisor | No human advisor can personally serve 1,000+ clients. Firms above this line are robo-advisers, app-based investing and managed-account platforms (Acorns, Betterment, Wealthfront, Human Interest, AssetMark). Among individual-focused firms the 99th percentile is ~775 clients per advisor, so 1,000 sits at the natural break |
| `investment_manager` (excluded) | ≥ 50% individual AUM, but no financial planning (Item 5G(1), and no planning clients in 5H) and does not select other advisers (5G(7)), and either: (a) ≥ 25% of AUM managed as portfolio manager in other firms' wrap programs (5I(2)(b)); or (b) > 200 clients per advisor with ≥ 10 advisors and ≥ <span>&#36;</span>250K AUM per client | Direct-indexing, SMA and model-portfolio providers used by RIAs (Quantinno, Vise AI, Aperio, Breckinridge, PGIM Custom Harvest, Vestmark). They report the end investors in other advisers' programs as their own individual clients, so Item 5D alone makes them look like wealth managers. Branch (a) is direct evidence of sub-advising. Branch (b) needs all three conditions, because client load alone also catches real advisory firms: mass-affluent and bank-channel advisers with small accounts (e.g. Nest Egg, ~<span>&#36;</span>47K per client) and one-person investment-counsel boutiques. Traditional investment counsel that simply don't offer planning (Sarofim Trust, Greenhaven) also stay wealth-focused |
| `family_office` (excluded) | ≥ 50% individual AUM, fewer than 20 clients and at least <span>&#36;</span>100M AUM | A handful of very wealthy families (median ~<span>&#36;</span>26M per client), e.g. BT Family Office, Fifth Avenue Family Office. High AUM per advisor here reflects a few large relationships rather than advisor strain, and these firms run a bespoke model rather than an advisor practice. Firms with under 20 clients and less than <span>&#36;</span>100M stay wealth-focused: they are small practices, kept out of the capacity ranking by its 20-client floor |
| `wealth_focused` | ≥ 50% of AUM from individuals + high-net-worth individuals (Item 5D(a)+(b)), and not one of the types above | Bespoke wealth managers: Nevis's core market and the focus of every ranking |
| `<50% individual AUM` | everything else: less than half of AUM from individuals + high-net-worth individuals | Mostly fund, ETF, pension and other institutional managers; shown in the all-RIA lists |

**Note on AUM totals:** RAUM is reported firm by firm, so assets one adviser manages as a sub-adviser for another are counted
by both; at least <span>&#36;</span>3.3T in 2026 is explicitly reported as wrap sub-advisory. Per-firm metrics are unaffected, but market-wide
totals and the segments' shares of AUM are somewhat overstated.

**Known edge cases:**
- Edelman Financial Engines combines human planners with a large 401(k) advice business, and is excluded by the
  1,000-client platform rule.
- Asset managers that also report financial planning (e.g. GW&K) stay wealth-focused.

## 6. Q1: Who are the fastest-growing firms?

Growth = **AUM CAGR** (compound annual growth rate) between the start snapshot and 2026:
- 1y: 2025→2026
- 3y: 2023→2026
- 5y: 2021→2026

Total growth over the window is shown too. Firms are grouped by their **2026** segment. The growth summary below
(median, 90th percentile, eligible firms) is shown for **wealth-focused RIAs** and for **all RIAs**, in separate, labeled tables.

To be eligible, a firm must:
- exist in both snapshots;
- have **≥ <span>&#36;</span>100M AUM at the start**;
- have a current (non-stale) filing in both years;
- have AUM above zero at the end.

`consolidation_flag` marks firms that became umbrella filers, added relying advisers, or report acquired firms. For these firms part of the growth is likely M&A rather than organic.

Each window shows the top 10 **wealth-focused** firms per segment (Nevis's core market), then the top 5 across **all RIAs**.
The full top-10 for both lists is in `outputs/q1_fastest_growing.csv`.

```python
growth = {h: ap.growth_table(panel, LATEST, h) for h in HORIZONS}

def eligible_growth(g):
    return g[(g["aum_start"] >= MIN_START_AUM) & ~g["is_stale"] & ~g["is_stale_start"] & (g["aum"] > 0)]

# Every aggregate table is shown for both universes, labeled, so the wealth-focused rankings can be read
# against the wealth-focused typical firm (and the market as a whole).
UNIVERSES = {"Wealth-focused RIAs": lambda d: d[d["wealth_focused"]], "All RIAs": lambda d: d}
windows = {f"{h}y": f"{h}-year" for h in HORIZONS}

def by_window(stats):
    """{label: Series indexed by (horizon, segment)} -> one table: rows = segment, columns = 'label, n-year'."""
    return pd.DataFrame({f"{label}, {windows[h]}": s.xs(h, level="horizon")
                         for label, s in stats.items() for h in windows})

growth_summary = pd.concat({(u, f"{h}y"): keep(eligible_growth(g)).groupby("segment", observed=False)["aum_cagr"]
                                          .agg(eligible_firms="count", median_cagr="median",
                                               p90_cagr=lambda s: s.quantile(0.9))
                            for u, keep in UNIVERSES.items() for h, g in growth.items()},
                           names=["universe", "horizon"])
growth_summary.to_csv(OUT / "q1_growth_summary.csv")
for u in UNIVERSES:
    gs = growth_summary.xs(u, level="universe")
    t = by_window({"Median CAGR": gs["median_cagr"], "90th-pct CAGR": gs["p90_cagr"], "Eligible firms": gs["eligible_firms"]})
    heading(f"AUM growth by segment: {u}", level=5)
    show(fmt(t, pct_cols=[c for c in t if "CAGR" in c], int_cols=[c for c in t if c.startswith("Eligible")]))
```

```python
Q1_COLS = ["sec_number", "firm_name", "city", "state", "aum_start", "aum", "aum_cagr", "aum_growth_total",
           "advisors_start", "advisors", "pct_aum_individuals", "consolidation_flag"]
q1 = []
for h, g in growth.items():
    heading(f"{h}-year window ({LATEST - h}→{LATEST})", level=3)
    q1.append(ranked_lists(eligible_growth(g), "aum_cagr", Q1_COLS, f"Fastest {h}-year AUM CAGR",
                           money_cols=["aum_start", "aum"], pct_cols=["aum_cagr", "aum_growth_total"],
                           share_cols=["pct_aum_individuals"], int_cols=["advisors_start", "advisors"],
                           flag_cols=["consolidation_flag"])
              .assign(horizon=f"{h}y"))
pd.concat(q1).to_csv(OUT / "q1_fastest_growing.csv", index=False)
```

#### Consistent growers
A single-window ranking rewards one-off jumps such as an acquisition or a large mandate win. These firms instead rank in the
**top quartile of their segment in all three windows**: sustained growth, a stronger signal of a firm scaling its operations.

As with the other rankings, there are two lists:
1. **Wealth-focused RIAs**, with quartiles calculated among wealth-focused peers only, so a wealth manager is compared with
   other wealth managers rather than with hedge funds;
2. **all RIAs**, with quartiles among all eligible firms.

Firms are sorted by 5-year AUM CAGR. The full lists are in `outputs/q1_consistent_growers.csv`.

```python
def consistent_growers(universe):
    """Firms in the top quartile of their segment's AUM CAGR in every window. `universe` filters the
    eligible firms before ranking, so quartiles are relative to that peer group."""
    ranks = []
    for h, g in growth.items():
        e = universe(eligible_growth(g))
        ranks.append(e.assign(**{f"pctile_{h}y": e.groupby("segment", observed=True)["aum_cagr"].rank(pct=True)})
                      .set_index("crd")[[f"pctile_{h}y"]])
    pct_cols = [f"pctile_{h}y" for h in HORIZONS]
    c = latest.set_index("crd").join(pd.concat(ranks, axis=1), how="inner").dropna(subset=pct_cols)
    c = c[c[pct_cols].min(axis=1) >= 0.75]
    return c.join(growth[5].set_index("crd")[["aum_cagr", "aum_start"]]).reset_index()

CONSISTENT_COLS = ["firm_name", "city", "state", "aum_start", "aum", "aum_cagr", "advisors", "pct_aum_individuals"]
consistent = {"Wealth-focused RIAs": consistent_growers(lambda d: d[d["wealth_focused"]]),
              "All RIAs": consistent_growers(lambda d: d)}
for label, c in consistent.items():
    print(f"Consistent top-quartile growers, {label}: {len(c):,}  ({per_segment(c.groupby('segment', observed=False).size())})")
for (label, c), n_show in zip(consistent.items(), [TOP_N, 5]):
    note = "" if n_show == TOP_N else f" (top {n_show} shown)"
    heading(f"Consistent growers: {label}{note}")
    top = ap.top_n_by_segment(c, "aum_cagr", n_show, CONSISTENT_COLS)
    show(fmt(top.set_index(["segment", "rank"]).rename(columns={"aum_start": "aum_2021", "aum_cagr": "5-year AUM CAGR"}),
             money_cols=["aum_2021", "aum"], pct_cols=["5-year AUM CAGR"], share_cols=["pct_aum_individuals"],
             int_cols=["advisors"]))
pd.concat([c.assign(universe=label) for label, c in consistent.items()]).to_csv(OUT / "q1_consistent_growers.csv", index=False)
```

**Commentary: fastest growing**
- **The typical RIA grows at roughly market returns; wealth-focused firms grow faster.**
  - **All RIAs:** median AUM CAGR is 13–15% over 1 and 3 years and 8–12% over 5 years.
  - **Wealth-focused RIAs:** 15–24% and 10–18%, highest in the larger segments.

  The 90th percentile runs at 20–60%/yr in both universes, so a firm in the top lists is growing several times faster than its peers.
- **Across all RIAs, the leaders are mostly new funds and launches.** Hedge funds, CLO managers and ETF sponsors grow from the <span>&#36;</span>100M floor to billions.
  This is real growth but not a wealth-management signal, which is why the wealth-focused lists lead.
- **Wealth-focused leaders fall into three groups:**
  - tech-enabled advisers that still serve clients through human advisors (Savvy, Farther);
  - roll-up platforms (Focus Partners Wealth, &Partners, Osaic Wealth, Cetera, Savant, Merit Financial);
  - fast-scaling independents (Maia Wealth, Lifeworks, Wedmont, Greenup Street).

  Two other groups would also rank here but are excluded:
  - robo-advisers and app-based investing firms (e.g. Robinhood Asset Management, Frec), classified as digital platforms
    (> 1,000 clients per advisor);
  - direct-indexing and SMA providers used by RIAs (e.g. Quantinno, Vise AI), classified as investment managers.

  The first two groups show that technology and M&A are the main growth engines in wealth management. Firms competing
  with them have a reason to invest in productivity.
- **Consistency matters.** 546 wealth-focused firms rank in the top quartile of their wealth-focused peers in all three
  windows: 364 under <span>&#36;</span>1B, 156 at <span>&#36;</span>1B–<span>&#36;</span>10B, 20 at <span>&#36;</span>10B–<span>&#36;</span>50B and 6 above <span>&#36;</span>50B. Across all RIAs the figure is 996. These are firms

## 7. Q2: Which firms are capacity-strained?

**AUM per advisor** (AUM ÷ 5B(1) advisory staff) in 2026. A firm qualifies with ≥ 2 advisors and a current filing.

Raw AUM per advisor is highest at institutional managers: fund and ETF sponsors where a small team runs billions.
These firms are capacity-*rich* by design, not strained. Digital platforms (robo-advisers, app investing) show the same effect
in client terms, with tens or hundreds of thousands of clients per advisor. So each segment gets two lists:
1. **wealth-focused RIAs** with **at least 20 clients**: bespoke wealth managers, excluding digital platforms, investment
   managers and family offices (Section 2.1). Here high AUM per advisor means each advisor carries a heavy client book, which
   is the strain relevant to Nevis. The 20-client floor keeps out firms whose high AUM per advisor comes from a handful of
   very large relationships;
2. all RIAs, for completeness.

`clients_per_advisor` is shown alongside as a second view of the load on each advisor.

#### Outlier handling: platform exclusion vs winsorizing
Clients per advisor is extremely skewed: Acorns reports ~580,000. The table compares four ways of averaging it across
individual-focused firms (≥ 50% of AUM from individuals, *including* digital platforms):
- the raw mean;
- the mean **winsorized** at the 1st/99th percentile within each segment (extremes clipped to the percentile, not dropped);
- the mean with **digital platforms excluded**;
- both together, and the median for reference.

```python
def winsorized_mean(s, lower=0.01, upper=0.99):
    s = s.dropna()
    return s.clip(s.quantile(lower), s.quantile(upper)).mean()

indiv = cap[cap["pct_aum_individuals"] >= WEALTH_SHARE]
bespoke = indiv[~indiv["digital_platform"]]
rows = []
for metric in ["clients_per_advisor", "aum_per_advisor"]:
    for s in ap.SEGMENT_LABELS:
        a, b = indiv.loc[indiv["segment"] == s, metric], bespoke.loc[bespoke["segment"] == s, metric]
        rows.append({"metric": metric, "segment": s, "Mean (raw)": a.mean(),
                     "Mean, winsorized 1–99%": winsorized_mean(a),
                     "Mean, platforms excluded": b.mean(),
                     "Mean, excluded + winsorized": winsorized_mean(b), "Median, excluded": b.median()})
outlier_cmp = pd.DataFrame(rows).set_index(["metric", "segment"])
stat_cols = list(outlier_cmp.columns)
heading("Clients per advisor: individual-focused firms (≥ 50% individual AUM, incl. digital platforms)", level=5)
show(fmt(outlier_cmp.loc["clients_per_advisor"], int_cols=stat_cols))
heading("AUM per advisor: individual-focused firms (≥ 50% individual AUM, incl. digital platforms)", level=5)
show(fmt(outlier_cmp.loc["aum_per_advisor"], money_cols=stat_cols))
```

Winsorizing works for the &lt;<span>&#36;</span>1B segment, but it is not enough at <span>&#36;</span>10B and above. Those segments have few individual-focused firms,
so the 99th percentile is itself a platform. In the &gt;<span>&#36;</span>50B segment the winsorized mean is still ~3,400 clients per advisor, against 134 once
platforms are excluded. Excluding platforms by rule removes the problem at its
source, and also takes them out of the rankings, which winsorizing cannot do.

**Commentary: capacity-strained**
- **There is a human capacity ceiling.** Among wealth-focused firms, median AUM per advisor rises from <span>&#36;</span>71M (&lt;<span>&#36;</span>1B) to ~<span>&#36;</span>146M

  advisors, not by raising per-advisor productivity. This is the constraint an AI platform like Nevis attacks: lifting this
  plateau is the value proposition, and the gap between a firm and its segment median sizes the opportunity.
- **Institutional managers are a different business.** Above <span>&#36;</span>10B, firms with mostly institutional AUM reach a median of
  <span>&#36;</span>0.6–1.2B per advisor with 1–2 clients each (funds and mandates). Their high ratios reflect the business model, not strain, so they should not be treated as prospects.
- **Digital platforms are a third business model.** 115 firms serve more than 1,000 clients per advisor: 25.6M clients and <span>&#36;</span>7.3T

- **Investment managers used by RIAs are a fourth.** 52 direct-indexing, SMA and model-portfolio providers (<span>&#36;</span>450B) look

  flagged 164 firms, but also caught real advisory firms such as mass-affluent and bank-channel advisers (e.g. Nest Egg) and
  one-person investment-counsel boutiques. Those stay wealth-focused.
- **Family offices are a fifth.** 163 individual-focused firms (<span>&#36;</span>93B) serve fewer than 20 clients each, with a median of 11
  clients and ~<span>&#36;</span>26M per client. Their AUM per advisor can be extreme, e.g. H&H International at <span>&#36;</span>12.7B per advisor across 6
  clients, but that reflects a few very large relationships, not strain. They are excluded from the wealth-focused lists, and
  the capacity ranking also requires at least 20 clients.
- **The most actionable names** are **traditional advisory firms with a handful of advisors carrying <span>&#36;</span>1B+ each**, such as
  Provident Trust, Prowell Financial, Brave Warrior Advisors and Greenhaven Associates.
- **Robustness.** Switching the advisor count from 5B(1) to 5B(3) barely changes the wealth-focused ranking: rank correlation
  is 0.92–0.96, and 5–8 of each segment's top 10 stay the same. Across all RIAs the correlation is lower (0.79–0.95) and the top

  conclusions hold either way; individual rankings are best used as a prospect screen.

## 8. Q3: Which RIAs appear under-staffed?

**Growth gap = AUM CAGR − advisor CAGR** over the same window. A firm is *under-staffed* when AUM grew (CAGR > 0) faster than its
advisor headcount (gap > 0), so each advisor is carrying more assets than before.

To be eligible, a firm must:
- have ≥ <span>&#36;</span>100M AUM and ≥ 2 advisors at the start;
- have ≥ 1 advisor at the end;
- have current filings in both years.

Firms are ranked by the gap within their 2026 segment.

Because markets alone lift AUM about 10%/yr while headcount is sticky, almost any firm can show *some* gap. The segment shares are
therefore reported for any gap and for a **material gap of ≥ 10 pts/yr**. Each is shown for **wealth-focused RIAs** and for
**all RIAs**, in separate, labeled tables.

```python
def eligible_staffing(g):
    return g[(g["aum_start"] >= MIN_START_AUM) & (g["advisors_start"] >= MIN_ADVISORS) & (g["advisors"] >= 1)
             & ~g["is_stale"] & ~g["is_stale_start"]]

under = {h: eligible_staffing(g).assign(
             understaffed=lambda d: (d["growth_gap"] > 0) & (d["aum_cagr"] > 0),
             material=lambda d: (d["growth_gap"] >= MATERIAL_GAP) & (d["aum_cagr"] > 0))
         for h, g in growth.items()}
gap_shares = pd.concat({
    (uname, f"{h}y"): keep(df).groupby("segment", observed=False).agg(
        any_gap_share=("understaffed", "mean"), material_gap_share=("material", "mean"),
        material_gap_firms=("material", "sum"))
    for uname, keep in UNIVERSES.items() for h, df in under.items()}, names=["universe", "horizon"])
gap_shares.to_csv(OUT / "q3_gap_shares.csv")
for uname in UNIVERSES:
    gs = gap_shares.xs(uname, level="universe")
    t = by_window({"Any gap": gs["any_gap_share"], f"Gap ≥ {MATERIAL_GAP * 100:.0f} pts": gs["material_gap_share"],
                   "Firms with gap ≥ 10 pts": gs["material_gap_firms"]})
    heading(f"Share of eligible firms where AUM outgrew advisor headcount: {uname}", level=5)
    show(fmt(t, share_cols=[c for c in t if not c.startswith("Firms")], int_cols=[c for c in t if c.startswith("Firms")]))
# the chart shows the wealth-focused universe, the focus of the rankings
material = gap_shares.xs("Wealth-focused RIAs", level="universe")["material_gap_share"].unstack("horizon")
```

```python
fig, ax = plt.subplots(figsize=(9, 3.4))
x = np.arange(len(material.index))
w = 0.26
for i, (col, color) in enumerate(zip(material.columns, [BLUE, ORANGE, AQUA])):
    bars = ax.bar(x + (i - 1) * (w + 0.02), material[col], width=w, color=color, label=f"{windows[col]} window")
    ax.bar_label(bars, labels=[f"{v:.0%}" for v in material[col]], fontsize=8, color=INK, padding=2)
ax.set_xticks(x, material.index.astype(str))
ax.set_ylim(0, max(0.5, material.values.max() * 1.25))
ax.yaxis.set_major_formatter(lambda v, _: f"{v:.0%}")
ax.grid(axis="x", visible=False)
ax.legend(loc="upper left", ncol=3)
ax.set_title(f"Wealth-focused RIAs: share whose AUM CAGR beat advisor CAGR by ≥ {MATERIAL_GAP * 100:.0f} pts/yr",
             loc="left")
plt.tight_layout()
plt.show()
```

#### Persistently under-staffed
Firms with a **material** gap (≥ 10 pts/yr) in **all three** windows. This is a structural gap rather than a one-year blip, and these

segment come first, then the top 5 across all RIAs, each sorted by 5-year gap. The full list is in

at the start of each window. For example, Sapere Wealth Management ranks #2 on the 5-year gap (wealth-focused, &lt;<span>&#36;</span>1B), but it had
only one advisor in 2023 and 2025, so its 1-year and 3-year gaps can't be measured. Its 5-year gap also comes mostly from a one-off
drop from 4 advisors to 1 in 2022.

```python
persist_ids = set.intersection(*[set(u.loc[u["material"], "crd"]) for u in under.values()])
persist = under[5][under[5]["crd"].isin(persist_ids)]
pw = persist[persist["wealth_focused"]]
print("Persistently under-staffed, all RIAs:      ", per_segment(persist.groupby("segment", observed=False).size()))
print("Persistently under-staffed, wealth-focused:", per_segment(pw.groupby("segment", observed=False).size()))
PERSIST_COLS = ["sec_number", "firm_name", "city", "state", "aum", "5-year AUM CAGR", "advisors_2021", "advisors",
                "5-year gap", "aum_per_advisor", "pct_aum_individuals"]
ranked_lists(persist.rename(columns={"aum_cagr": "5-year AUM CAGR", "growth_gap": "5-year gap",
                                     "advisors_start": "advisors_2021"}),
             "5-year gap", PERSIST_COLS, "Persistently under-staffed, ranked by 5-year gap", csv_note="full list in CSV",
             money_cols=["aum", "aum_per_advisor"], pct_cols=["5-year AUM CAGR", "5-year gap"],
             share_cols=["pct_aum_individuals"], int_cols=["advisors_2021", "advisors"])
persist.to_csv(OUT / "q3_persistently_understaffed.csv", index=False)
```

**Commentary: under-staffed**
- **Under-staffing is the norm, mostly because of markets.**
  - **All RIAs:** 68–80% of eligible firms grew AUM faster than advisors, depending on segment and window. With the
    10-pts/yr materiality bar the share drops to ~50% over 1–3 years and **34–39% over 5 years**, still a third of the industry.
  - **Wealth-focused RIAs are more stretched:** 78–91% show some gap, and **60–76%** over 1–3 years and **41–54%** over
    5 years show a material one.

  Wealth-focused firms grow AUM faster (Q1), but the median firm in both groups kept advisor headcount flat over every window,
  so each wealth advisor carries more every year.
- **Persistent gaps are the strongest signal.** 2,103 firms have a material gap in all three windows, 1,339 of them
  wealth-focused (914 under <span>&#36;</span>1B, 366 at <span>&#36;</span>1B–<span>&#36;</span>10B). Their assets compounded while headcount stood still, for five years.
  These firms most need advisor-productivity tooling and are the highest-intent prospects for Nevis.
  `outputs/q3_persistently_understaffed.csv` is the ready-made list.
- **Large firms show it too.** Above <span>&#36;</span>50B, consolidators such as Osaic and Cetera grew advisor headcount 3–4x yet still saw AUM
  grow faster. Scale alone does not solve advisor capacity.
- **Caveat.** A fall in 5B(1) can be a re-classification (e.g. moving staff to an affiliate) rather than attrition. Large
  single-firm gaps, such as advisors falling from 127 to 14 at Acorns, should be checked before outreach.

### Bringing it together for Nevis
1. **Core market:** ~1,410 wealth-focused RIAs at <span>&#36;</span>1B–<span>&#36;</span>10B. Scaled enough to buy software, independent enough to decide fast.
   This is the fastest-growing segment by firm count.
2. **Prioritize inside it** with the three signals:
   - consistent top-quartile growth (Q1);
   - AUM per advisor above the ~<span>&#36;</span>150M plateau (Q2);
   - a persistent material staffing gap (Q3).
   Firms meeting two or more are the warmest leads.
3. **Long tail:** ~7,300 sub-<span>&#36;</span>1B wealth firms suit a lower-touch, product-led motion.
4. **Watch list:**
   - the 115 digital platforms (`outputs/digital_platforms.csv`), which serve 25.6M clients with ~3,000 advisors;
   - the 52 investment managers used by RIAs (`outputs/investment_managers.csv`), potential partners or channels;
   - the tech-enabled advisers in the growth lists.

   They show what technology does to the advisor-capacity ceiling, and they are the competitive benchmark Nevis's customers face.
